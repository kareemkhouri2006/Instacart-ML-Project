### import libraries

```python
import pandas as pd
```

### Load datasets (Relative Path)


```python
aisles=pd.read_csv("../data/raw/aisles.csv")
departments=pd.read_csv("../data/raw/departments.csv")
order_products_prior=pd.read_csv("../data/raw/order_products__prior.csv")
order_products_train=pd.read_csv("../data/raw/order_products__train.csv")
orders=pd.read_csv("../data/raw/orders.csv")
products=pd.read_csv("../data/raw/products.csv")
```

### Memory-savvy

```python
orders["order_id"] = orders["order_id"].astype("int32")
orders["user_id"] = orders["user_id"].astype("int32")
orders["order_number"] = orders["order_number"].astype("int16")
orders["order_dow"] = orders["order_dow"].astype("int8")
orders["order_hour_of_day"] = orders["order_hour_of_day"].astype("int8")
orders["days_since_prior_order"] = orders["days_since_prior_order"].astype("float32")
```

```python
products["product_id"] = products["product_id"].astype("int32")
products["aisle_id"] = products["aisle_id"].astype("int16")
products["department_id"] = products["department_id"].astype("int16")
```

```python
order_products_prior["order_id"] = order_products_prior["order_id"].astype("int32")
order_products_prior["product_id"] = order_products_prior["product_id"].astype("int32")
order_products_prior["add_to_cart_order"] = order_products_prior["add_to_cart_order"].astype("int16")
order_products_prior["reordered"] = order_products_prior["reordered"].astype("int8")

order_products_train["order_id"] = order_products_train["order_id"].astype("int32")
order_products_train["product_id"] = order_products_train["product_id"].astype("int32")
order_products_train["add_to_cart_order"] = order_products_train["add_to_cart_order"].astype("int16")
order_products_train["reordered"] = order_products_train["reordered"].astype("int8")
```

```python
aisles["aisle"] = aisles["aisle"].astype("category")
departments["department"] = departments["department"].astype("category")
products["product_name"] = products["product_name"].astype("category")
```

### join 

```python
df1 = order_products_prior.merge(orders,on="order_id",how="left")
df2 = df1.merge(products,on="product_id",how="left")
df3 = df2.merge(aisles,on="aisle_id",how="left")
df_prior = df3.merge(departments,on="department_id",how="left")
```

```python
df4 = order_products_train.merge(orders,on="order_id",how="left")
df5 = df4.merge(products,on="product_id",how="left")
df6 = df5.merge(aisles,on="aisle_id",how="left")
df_train = df6.merge(departments,on="department_id",how="left")
```
